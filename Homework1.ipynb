```python
# Bài 5
c = input("Nhập vào một chữ cái hoa: ")

if 'A' <= c <= 'Z':
    vi_tri_moi = (ord(c) - ord('A') - 1) % 26
    result = chr(ord('a') + vi_tri_moi)
    
    print(f"Ký tự liền trước chữ thường tương ứng là: {result}")
else:
    print("Ký tự nhập vào không phải chữ cái hoa!")

```
